```python
# Without a function — you have to rewrite this for every sensor
turbidity = [20, 85, 142, 67, 110]
min_t = min(turbidity)
max_t = max(turbidity)
normalized_turbidity = [(x - min_t) / (max_t - min_t) for x in turbidity]
print(normalized_turbidity)

# With a function — write once, use forever
def normalize(data):
    min_val = min(data)
    max_val = max(data)
    return [(x - min_val) / (max_val - min_val) for x in data]

# Now use it for ANY sensor with one line
turbidity_norm = normalize([20, 85, 142, 67, 110])
chemical_norm  = normalize([5, 50, 120, 30, 200])
ph_norm        = normalize([6.5, 3.2, 8.9, 7.1, 4.0])

print(turbidity_norm)
print(chemical_norm)
print(ph_norm)
```

```python
def is_polluted(ph, turbidity):
    # Assuming pollution if pH is too low or turbidity is too high
    return ph < 6.5 or turbidity > 100

print(is_polluted(4.5, 95))   # True  — pH is low
print(is_polluted(7.0, 50))   # False — both normal
print(is_polluted(7.0, 110))  # True  — turbidity high
```
